```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings
import joblib

from sklearn.model_selection import train_test_split, GridSearchCV, KFold
from sklearn.linear_model import RidgeCV
from sklearn.ensemble import RandomForestRegressor, HistGradientBoostingRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.inspection import permutation_importance, PartialDependenceDisplay
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tools import add_constant
import statsmodels.api as sm
from statsmodels.stats.diagnostic import het_breuschpagan, het_white

warnings.filterwarnings("ignore", category=RuntimeWarning, module="statsmodels")

RANDOM_STATE = 42
```

```python
df = pd.read_csv("Ukraine_flats_ML.csv")
print(df.shape)
print("Пропусків (очікувано -- підуть у Pipeline, ЛИШЕ на train-фолді):")
print(df.isna().sum()[df.isna().sum() > 0])
```

```python
CAT_PREFIX = "city_"
CAT_COL = "city_raw"
UKRAINE_CITIES = [
    "Lviv", "Kharkiv", "Dnipro", "Zaporizhzhia", "Odesa", "Poltava", "Mykolaiv",
    "Cherkasy", "Kropyvnytskyi", "Sumy", "Chernihiv", "Zhytomyr", "Lutsk",
    "Ivano-Frankivsk", "Ternopil", "Khmelnytskyi", "Uzhhorod", "Vinnytsia",
    "Rivne", "Chernivtsi", "Kherson",
]


def reconstruct_categorical(df: pd.DataFrame, prefix: str, master_list: list[str]) -> tuple[pd.Series, list[str]]:
    dummy_cols = [c for c in df.columns if c.startswith(prefix)]
    present = {c[len(prefix):] for c in dummy_cols}
    missing_sorted = sorted(m for m in master_list if m not in present)
    reference = missing_sorted[0]
    cat = pd.Series(reference, index=df.index, dtype="object")
    for c in dummy_cols:
        name = c[len(prefix):]
        cat = cat.where(~df[c].astype(bool), name)
    return cat, dummy_cols


df[CAT_COL], _dummy_cols = reconstruct_categorical(df, CAT_PREFIX, UKRAINE_CITIES)
df = df.drop(columns=_dummy_cols)
print(df[CAT_COL].value_counts())
```

```python
TARGET = "price"


def build_feature_frame(df: pd.DataFrame):
    """X лишається "сирим" (може містити NaN) -- жодної імпутації тут, вона
    відбувається ВСЕРЕДИНІ sklearn.Pipeline, fit_transform лише на
    train-фолді (усередині fit_variant нижче) -- усуває витік даних.
    ВЕСЬ X одразу float64 (не лише те, що pandas сам розпізнав як dtype==
    "bool") -- тристейт-булеві колонки (has_gas, autonomy_* тощо) зберігаються
    з preprocessing.ipynb як object (мікс True/False/NaN), і select_dtypes
    ("bool") їх не бачить -- лишались би object і випадали з numeric_only=True
    розрахунків (напр. .median() у тесті на гетероскедастичність нижче)."""
    feature_cols = [c for c in df.columns if c not in (TARGET, CAT_COL)]
    X = df[feature_cols].astype(float)
    y = np.log1p(df[TARGET].astype(float))
    return X, y, df[CAT_COL]


def fit_target_encoding(cat_train: pd.Series, y_train: pd.Series, m: float = 10.0):
    global_mean = float(y_train.mean())
    stats = y_train.groupby(cat_train).agg(["mean", "count"])
    mapping = (stats["count"] * stats["mean"] + m * global_mean) / (stats["count"] + m)
    return mapping, global_mean


def apply_target_encoding(cat: pd.Series, mapping: pd.Series, global_mean: float) -> pd.Series:
    return cat.map(mapping).fillna(global_mean)
```

```python
def drop_constant_columns(X: pd.DataFrame) -> pd.DataFrame:
    constant_cols = [c for c in X.columns if X[c].nunique(dropna=False) <= 1]
    if constant_cols:
        print(f"  Прибрано {len(constant_cols)} константних колонок: {constant_cols}")
    return X.drop(columns=constant_cols)


def prune_by_vif(X: pd.DataFrame, thresh: float = 10.0) -> list[str]:
    """VIF-відбір -- ЛИШЕ для Ridge (див. train_and_compare нижче). Ансамблі
    дерев (RandomForest/HistGradientBoosting) стійкі до мультиколінеарності:
    для дерев відбір за VIF тільки прибирає потенційно корисні ознаки без
    жодної користі для якості прогнозу, тож для них цей крок вимкнено."""
    X = drop_constant_columns(X).astype(float)
    X_const = add_constant(X, has_constant="add")

    while X_const.shape[1] > 2:
        vifs = pd.Series(
            [variance_inflation_factor(X_const.values, i) for i in range(X_const.shape[1])],
            index=X_const.columns,
        ).drop("const")
        worst = vifs.idxmax()
        if vifs[worst] < thresh:
            break
        print(f"  Дропаю '{worst}' (VIF={vifs[worst]:.1f})")
        X_const = X_const.drop(columns=[worst])

    kept = [c for c in X_const.columns if c != "const"]
    print(f"  Лишилось {len(kept)} ознак після VIF-відбору")
    return kept


def evaluate(y_true_log: np.ndarray, y_pred_log: np.ndarray) -> dict:
    y_true, y_pred = np.expm1(y_true_log), np.expm1(y_pred_log)
    return {
        "R2": r2_score(y_true_log, y_pred_log),
        "MAE_$/m2": mean_absolute_error(y_true, y_pred),
        "RMSE_$/m2": mean_squared_error(y_true, y_pred) ** 0.5,
        "MAPE_%": (np.abs((y_true - y_pred) / y_true).mean()) * 100,
    }


# Дискретні лічильники -- заповнюємо МОДОЮ (most_frequent), а не медіаною.
COUNT_COLS = ["bedroom_count", "toilets_count", "balcony_count"]


def build_pipeline(model_name: str, model, columns: list[str]) -> Pipeline:
    """columns -- фактичний список ознак цього виклику (full_cols / kept_cols
    після VIF / фолд у CV), бо ColumnTransformer потребує знати наперед, які
    саме колонки де саме є."""
    steps = []
    if model_name in ("Ridge", "RandomForest"):
        count_cols_present = [c for c in COUNT_COLS if c in columns]
        rest_cols_present = [c for c in columns if c not in count_cols_present]
        transformers = []
        if count_cols_present:
            transformers.append(("count_mode", SimpleImputer(strategy="most_frequent"), count_cols_present))
        if rest_cols_present:
            transformers.append(("rest_median", SimpleImputer(strategy="median"), rest_cols_present))
        steps.append(("imputer", ColumnTransformer(transformers, remainder="drop")))
    if model_name == "Ridge":
        steps.append(("scaler", StandardScaler()))
    steps.append(("model", model))
    return Pipeline(steps)


RF_GRID = {
    "model__n_estimators": [200, 400, 600],
    "model__max_depth": [None, 8, 12, 20],
    "model__min_samples_leaf": [1, 2, 4],
}
HGB_GRID = {
    "model__max_iter": [200, 400, 600],
    "model__max_depth": [None, 4, 8],
    "model__learning_rate": [0.03, 0.06, 0.1],
}


def fit_variant(model_name, X_train, X_test, y_train, y_test, cols, fixed_params=None):
    """Повертає ФІТОВАНИЙ Pipeline -- pipeline.predict(X) сам імпутує/
    масштабує вхід так само, як під час навчання."""
    X_train_c, X_test_c = X_train[cols], X_test[cols]

    if model_name == "Ridge":
        ridge = RidgeCV(alphas=[fixed_params["alpha_"]]) if fixed_params is not None \
            else RidgeCV(alphas=np.logspace(-2, 3, 30))
        pipeline = build_pipeline(model_name, ridge, cols).fit(X_train_c, y_train)

    elif model_name == "RandomForest":
        if fixed_params is not None:
            rf = RandomForestRegressor(random_state=RANDOM_STATE, **fixed_params)
            pipeline = build_pipeline(model_name, rf, cols).fit(X_train_c, y_train)
        else:
            base = build_pipeline(model_name, RandomForestRegressor(random_state=RANDOM_STATE), cols)
            search = GridSearchCV(base, RF_GRID, cv=5, scoring="r2", n_jobs=-1)
            search.fit(X_train_c, y_train)
            pipeline = search.best_estimator_

    else:  # HistGradientBoosting
        if fixed_params is not None:
            hgb = HistGradientBoostingRegressor(random_state=RANDOM_STATE, **fixed_params)
            pipeline = build_pipeline(model_name, hgb, cols).fit(X_train_c, y_train)
        else:
            base = build_pipeline(model_name, HistGradientBoostingRegressor(random_state=RANDOM_STATE), cols)
            search = GridSearchCV(base, HGB_GRID, cv=5, scoring="r2", n_jobs=-1)
            search.fit(X_train_c, y_train)
            pipeline = search.best_estimator_

    metrics = evaluate(y_test.values, pipeline.predict(X_test_c))
    return pipeline, X_test_c, metrics


def get_model_params(model_name: str, pipeline: Pipeline) -> dict:
    model = pipeline.named_steps["model"]
    if model_name == "Ridge":
        return {"alpha_": model.alpha_}
    if model_name == "RandomForest":
        return {k: v for k, v in model.get_params().items()
                if k in ("n_estimators", "max_depth", "min_samples_leaf")}
    return {k: v for k, v in model.get_params().items()
            if k in ("max_iter", "max_depth", "learning_rate")}
```

```python
def train_and_compare(df: pd.DataFrame, vif_thresh: float = 10.0):
    X, y, cat_raw = build_feature_frame(df)
    idx_train, idx_test = train_test_split(df.index, test_size=0.2, random_state=RANDOM_STATE)

    X_train, X_test = X.loc[idx_train].copy(), X.loc[idx_test].copy()
    y_train, y_test = y.loc[idx_train], y.loc[idx_test]

    te_mapping, te_global_mean = fit_target_encoding(cat_raw.loc[idx_train], y_train)
    enc_col = f"{CAT_COL}_target_enc"
    X_train[enc_col] = apply_target_encoding(cat_raw.loc[idx_train], te_mapping, te_global_mean).values
    X_test[enc_col] = apply_target_encoding(cat_raw.loc[idx_test], te_mapping, te_global_mean).values

    full_cols = list(X.columns) + [enc_col]

    print("--- Навчання на ПОВНОМУ наборі ознак ---")
    results, fitted = {}, {}
    for name in ["Ridge", "RandomForest", "HistGradientBoosting"]:
        pipeline, _, metrics = fit_variant(name, X_train, X_test, y_train, y_test, full_cols)
        results[name] = metrics
        fitted[name] = pipeline

    results_df = pd.DataFrame(results).T
    print("\n--- Результати на test (повний набір ознак) ---")
    print(results_df)

    best_name = results_df["MAPE_%"].idxmin()
    print(f"\n>>> Найкраща модель (за мінімальним MAPE_%): {best_name}")

    if best_name == "Ridge":
        print("\n--- VIF-відбір (лише для Ridge) ---")
        imputer_for_vif = SimpleImputer(strategy="median").fit(X_train)
        X_train_imputed = pd.DataFrame(
            imputer_for_vif.transform(X_train), columns=X_train.columns, index=X_train.index,
        )
        kept_cols = prune_by_vif(X_train_imputed, thresh=vif_thresh)
        best_model_pruned, X_test_pruned, pruned_metrics = fit_variant(
            best_name, X_train, X_test, y_train, y_test, kept_cols
        )
        print(f"\nMAPE_% {best_name}: повний набір ознак = {results_df.loc[best_name, 'MAPE_%']:.1f}%, "
              f"після VIF-відбору = {pruned_metrics['MAPE_%']:.1f}%")
    else:
        print(f"\n--- VIF-відбір ПРОПУЩЕНО: {best_name} -- дерево, стійке до мультиколінеарності ---")
        kept_cols = full_cols
        best_model_pruned = fitted[best_name]
        X_test_pruned = X_test[kept_cols]

    perm = permutation_importance(
        best_model_pruned, X_test_pruned, y_test, n_repeats=20,
        random_state=RANDOM_STATE, scoring="r2", n_jobs=-1,
    )
    importance_df = (
        pd.DataFrame({"feature": kept_cols, "importance": perm.importances_mean,
                      "std": perm.importances_std})
        .sort_values("importance", ascending=False)
        .reset_index(drop=True)
    )

    top_n = importance_df.head(15)
    plt.figure(figsize=(8, 6))
    plt.barh(top_n["feature"][::-1], top_n["importance"][::-1], xerr=top_n["std"][::-1])
    plt.xlabel("Permutation importance (спад R² при перемішуванні ознаки)")
    plt.title(f"Квартири (Україна) — {best_name} — топ ознак, ціль = $/м²")
    plt.tight_layout()
    plt.show()

    return {
        "results_df": results_df,
        "best_name": best_name,
        "best_model_pruned": best_model_pruned,
        "fixed_params": get_model_params(best_name, best_model_pruned),
        "importance_df": importance_df,
        "kept_cols": kept_cols,
        "enc_col": enc_col,
    }


result = train_and_compare(df)
```

```python
def test_heteroscedasticity(df: pd.DataFrame, result: dict):
    """Breusch-Pagan і White тести + графіки залишків vs передбачення."""
    X, y, cat_raw = build_feature_frame(df)
    idx_train, idx_test = train_test_split(df.index, test_size=0.2, random_state=RANDOM_STATE)
    X_train, X_test = X.loc[idx_train].copy(), X.loc[idx_test].copy()
    y_train, y_test = y.loc[idx_train], y.loc[idx_test]

    te_mapping, te_global_mean = fit_target_encoding(cat_raw.loc[idx_train], y_train)
    enc_col = result["enc_col"]
    X_test[enc_col] = apply_target_encoding(cat_raw.loc[idx_test], te_mapping, te_global_mean).values

    kept_cols = result["kept_cols"]
    model = result["best_model_pruned"]  # повний Pipeline

    X_test_c = X_test[kept_cols]
    y_pred_log = model.predict(X_test_c)
    resid_log = y_test.values - y_pred_log

    y_true_price = np.expm1(y_test.values)
    y_pred_price = np.expm1(y_pred_log)
    resid_price = y_true_price - y_pred_price

    fig, axes = plt.subplots(1, 2, figsize=(13, 5))
    axes[0].scatter(y_pred_log, resid_log, s=12, alpha=0.4)
    axes[0].axhline(0, color="black", linewidth=1)
    axes[0].set_xlabel("Передбачення, log($/м²)")
    axes[0].set_ylabel("Залишок (лог-шкала)")
    axes[0].set_title("Залишки vs передбачення -- лог-шкала")

    axes[1].scatter(y_pred_price, resid_price, s=12, alpha=0.4)
    axes[1].axhline(0, color="black", linewidth=1)
    axes[1].set_xlabel("Передбачена ціна, $/м²")
    axes[1].set_ylabel("Залишок, $/м²")
    axes[1].set_title("Залишки vs передбачення -- відновлена ціна")
    plt.tight_layout()
    plt.show()

    X_test_for_test = X_test_c.astype(float)
    X_test_for_test = X_test_for_test.fillna(X_test_for_test.median(numeric_only=True))
    exog = sm.add_constant(X_test_for_test)
    bp_stat, bp_pvalue, bp_f_stat, bp_f_pvalue = het_breuschpagan(resid_log, exog)

    white_pvalue = None
    try:
        white_stat, white_pvalue, white_f_stat, white_f_pvalue = het_white(resid_log, exog)
    except (AssertionError, np.linalg.LinAlgError) as exc:
        print(f"White-тест не вдалося порахувати (типово через dummy-колінеарність): {exc!r}")

    print(f"Модель: {result['best_name']}, n_test={len(y_test)}, ознак={len(kept_cols)}\n")
    print("=== Breusch-Pagan (залишки на лог-шкалі) ===")
    print(f"LM-статистика={bp_stat:.2f}, p-value={bp_pvalue:.4f}")
    print(f"F-статистика={bp_f_stat:.2f}, p-value={bp_f_pvalue:.4f}")
    if white_pvalue is not None:
        print("\n=== White (залишки на лог-шкалі) ===")
        print(f"LM-статистика={white_stat:.2f}, p-value={white_pvalue:.4f}")
        print(f"F-статистика={white_f_stat:.2f}, p-value={white_f_pvalue:.4f}")

    alpha = 0.05
    is_hetero = (bp_pvalue < alpha) or (white_pvalue is not None and white_pvalue < alpha)
    verdict = "Є ознаки гетероскедастичності" if is_hetero else "Гетероскедастичність не виявлено"
    print(f"\n>>> {verdict} (α=0.05)")

    return {"resid_log": resid_log, "resid_price": resid_price, "bp_pvalue": bp_pvalue, "white_pvalue": white_pvalue}


hetero_result = test_heteroscedasticity(df, result)
```

```python
QUANTILES = [0.1, 0.5, 0.9]


def compute_conformity_quantile(conformity_scores: np.ndarray, target_coverage: float) -> float:
    """Conformalized Quantile Regression (CQR, Romano et al. 2019): точна
    формула для скінченної вибірки (ceil((n+1)*cov)/n, а не просто cov) --
    дає асимптотично точну гарантію покриття на незалежному test."""
    n = len(conformity_scores)
    level = min(1.0, np.ceil((n + 1) * target_coverage) / n)
    return float(np.quantile(conformity_scores, level))


def fit_quantile_models(df: pd.DataFrame, kept_cols: list[str], enc_col: str,
                         quantiles: list[float] = QUANTILES):
    """HistGradientBoostingRegressor(loss='quantile') працює напряму з NaN.
    Розбиває train на proper-train (60%) і calib (20%) -- калібрувальна
    вибірка йде на CQR-поправку, а не на навчання квантильних моделей."""
    X, y, cat_raw = build_feature_frame(df)
    idx_train, idx_test = train_test_split(df.index, test_size=0.2, random_state=RANDOM_STATE)
    idx_proper_train, idx_calib = train_test_split(idx_train, test_size=0.25, random_state=RANDOM_STATE)

    X_proper_train, X_calib, X_test = X.loc[idx_proper_train].copy(), X.loc[idx_calib].copy(), X.loc[idx_test].copy()
    y_proper_train, y_calib, y_test = y.loc[idx_proper_train], y.loc[idx_calib], y.loc[idx_test]

    te_mapping, te_global_mean = fit_target_encoding(cat_raw.loc[idx_proper_train], y_proper_train)
    X_proper_train[enc_col] = apply_target_encoding(cat_raw.loc[idx_proper_train], te_mapping, te_global_mean).values
    X_calib[enc_col] = apply_target_encoding(cat_raw.loc[idx_calib], te_mapping, te_global_mean).values
    X_test[enc_col] = apply_target_encoding(cat_raw.loc[idx_test], te_mapping, te_global_mean).values

    X_proper_train_c, X_calib_c, X_test_c = X_proper_train[kept_cols], X_calib[kept_cols], X_test[kept_cols]

    models = {}
    for q in quantiles:
        model = HistGradientBoostingRegressor(
            loss="quantile", quantile=q, random_state=RANDOM_STATE,
            max_iter=300, max_depth=8, learning_rate=0.06,
        )
        model.fit(X_proper_train_c, y_proper_train)
        models[q] = model

    q_lo, q_hi = min(quantiles), max(quantiles)
    target_coverage = q_hi - q_lo

    raw_low_test, raw_high_test = models[q_lo].predict(X_test_c), models[q_hi].predict(X_test_c)
    raw_coverage = ((y_test.values >= raw_low_test) & (y_test.values <= raw_high_test)).mean()

    calib_low, calib_high = models[q_lo].predict(X_calib_c), models[q_hi].predict(X_calib_c)
    conformity_scores = np.maximum(calib_low - y_calib.values, y_calib.values - calib_high)
    q_hat = compute_conformity_quantile(conformity_scores, target_coverage)

    calibrated_low_test = raw_low_test - q_hat
    calibrated_high_test = raw_high_test + q_hat
    calibrated_coverage = ((y_test.values >= calibrated_low_test) & (y_test.values <= calibrated_high_test)).mean()

    print(f"Очікуване покриття інтервалу [{q_lo}, {q_hi}]: {target_coverage:.0%}")
    print(f"Сире покриття (без калібрування) на test: {raw_coverage:.1%}")
    print(f"Q_hat (CQR-поправка, лог-шкала): {q_hat:.4f}")
    print(f"Каліброване покриття (CQR) на test: {calibrated_coverage:.1%}")

    price_low, price_high = np.expm1(calibrated_low_test), np.expm1(calibrated_high_test)
    price_median = np.expm1(models[0.5].predict(X_test_c))
    median_width_pct = float(np.median((price_high - price_low) / price_median * 100))
    print(f"Медіанна ширина каліброваного інтервалу: {median_width_pct:.1f}% від передбаченої ціни")

    return models, q_hat


quantile_models, quantile_q_hat = fit_quantile_models(df, result["kept_cols"], result["enc_col"])
```

```python
def cross_validate_best(df: pd.DataFrame, best_name: str, kept_cols: list[str],
                         fixed_params: dict, enc_col: str, n_splits: int = 5):
    X, y, cat_raw = build_feature_frame(df)
    kf = KFold(n_splits=n_splits, shuffle=True, random_state=RANDOM_STATE)
    fold_metrics = []

    for fold_i, (tr_idx, val_idx) in enumerate(kf.split(df.index), start=1):
        idx_tr = df.index[tr_idx]
        idx_val = df.index[val_idx]

        X_tr, X_val = X.loc[idx_tr].copy(), X.loc[idx_val].copy()
        y_tr, y_val = y.loc[idx_tr], y.loc[idx_val]

        te_mapping, te_global_mean = fit_target_encoding(cat_raw.loc[idx_tr], y_tr)
        X_tr[enc_col] = apply_target_encoding(cat_raw.loc[idx_tr], te_mapping, te_global_mean).values
        X_val[enc_col] = apply_target_encoding(cat_raw.loc[idx_val], te_mapping, te_global_mean).values

        cols = [c for c in kept_cols if c in X_tr.columns]
        _, _, metrics = fit_variant(best_name, X_tr, X_val, y_tr, y_val, cols, fixed_params=fixed_params)
        metrics["fold"] = fold_i
        fold_metrics.append(metrics)

    cv_df = pd.DataFrame(fold_metrics).set_index("fold")
    summary = cv_df.agg(["mean", "std"])
    print(cv_df)
    print("\n--- Підсумок по фолдах ---")
    print(summary)
    return cv_df


cv_results = cross_validate_best(
    df, result["best_name"], result["kept_cols"], result["fixed_params"], result["enc_col"],
)
```

```python
def save_production_pipeline(df: pd.DataFrame, best_name: str, kept_cols: list[str],
                             fixed_params: dict, enc_col: str, out_file: str,
                             quantile_models: dict, quantile_q_hat: float,
                             quantiles: list[float] = QUANTILES):
    """Артефакт зберігає ГОТОВИЙ Pipeline для ТОЧКОВОГО прогнозу -- він
    тренується на ВСІХ даних (для нього калібрування не потрібне).

    ВАЖЛИВО: quantile_models і quantile_q_hat -- НЕ рахуються тут заново, а
    приймаються як параметри з fit_quantile_models() вище. Раніше тут
    (помилково) перераховувались ЗАНОВО окремі квантильні моделі на іншому
    спліті (idx_calib випадково збігався з idx_test із діагностики через
    однаковий random_state) -- і в артефакт потрапляв Q_hat, чиє покриття
    НІКОЛИ не перевірялось (перевірене покриття ~80% належало іншій парі
    моделей/Q_hat, яка в продакшн не потрапляла). Тепер продакшн отримує
    РІВНО ТІ моделі й Q_hat, чиє покриття підтверджено вище."""
    X, y, cat_raw = build_feature_frame(df)

    te_mapping_full, te_global_mean_full = fit_target_encoding(cat_raw, y)
    X_full = X.copy()
    X_full[enc_col] = apply_target_encoding(cat_raw, te_mapping_full, te_global_mean_full).values
    cols = [c for c in kept_cols if c in X_full.columns]
    X_c, y_c = X_full[cols], y  # фінальна точкова модель -- на ВСІХ даних

    if best_name == "Ridge":
        model = build_pipeline("Ridge", RidgeCV(alphas=[fixed_params["alpha_"]]), cols).fit(X_c, y_c)
    elif best_name == "RandomForest":
        model = build_pipeline("RandomForest", RandomForestRegressor(random_state=RANDOM_STATE, **fixed_params), cols).fit(X_c, y_c)
    else:
        model = build_pipeline("HistGradientBoosting", HistGradientBoostingRegressor(random_state=RANDOM_STATE, **fixed_params), cols).fit(X_c, y_c)

    artifact = {
        "model_name": best_name,
        "model": model,  # повний Pipeline
        "feature_cols": cols,
        "cat_col": CAT_COL,
        "cat_prefix": CAT_PREFIX,
        "target_encoding_mapping": {str(k): float(v) for k, v in te_mapping_full.items()},
        "target_encoding_global_mean": te_global_mean_full,
        "target_enc_col": enc_col,
        "quantile_models": quantile_models,
        "quantiles": quantiles,
        "quantile_q_hat": quantile_q_hat,
    }
    joblib.dump(artifact, out_file)
    print(f"Збережено: {out_file}")
    return artifact


def predict_new_object(artifact: dict, new_row: dict) -> dict:
    """new_row — dict з сирими значеннями ознак (як у df ДО one-hot/target
    encoding), включно з CAT_COL (напр. 'city_raw'). Повертає dict з медіанною
    ціною за м² і, якщо є квантильні моделі, каліброваним (CQR) діапазоном."""
    row = pd.Series(new_row)
    cat_value = row.get(artifact["cat_col"], None)
    mapping = artifact["target_encoding_mapping"]
    row[artifact["target_enc_col"]] = mapping.get(cat_value, artifact["target_encoding_global_mean"])

    X_new = pd.DataFrame([row])[artifact["feature_cols"]].astype(float)
    pred_log = artifact["model"].predict(X_new)[0]  # pipeline сам імпутує/масштабує
    out = {"price_per_sqm": float(np.expm1(pred_log))}

    quantile_models = artifact.get("quantile_models", {})
    q_hat = artifact.get("quantile_q_hat", 0.0)
    if quantile_models:
        q_lo, q_hi = min(quantile_models), max(quantile_models)
        for q, qmodel in quantile_models.items():
            q_pred_log = qmodel.predict(X_new)[0]
            if q == q_lo:
                q_pred_log -= q_hat
            elif q == q_hi:
                q_pred_log += q_hat
            out[f"price_per_sqm_q{q}"] = float(np.expm1(q_pred_log))

    return out


artifact = save_production_pipeline(
    df, result["best_name"], result["kept_cols"], result["fixed_params"], result["enc_col"],
    out_file="flats_ukraine_model_pipeline.joblib",
    quantile_models=quantile_models, quantile_q_hat=quantile_q_hat,
)
```
